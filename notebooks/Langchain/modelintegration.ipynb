```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ['OPENAI_API_KEY'] =os.getenv("OPENAI_API_KEY")

```

```python
from langchain.chat_models import init_chat_model
models = init_chat_model("gpt-3.5-turbo")
models
```

```python
## invoke model

responce = models.invoke("What is the capital of France?")
responce.content
```

## streaming and batch

```python
## streaming and batch

models.invoke("What is AI?")
```

```python
for chunk in models.stream("What is AI in 200 words?"):
    print(chunk.content,end="|",flush=True)
```

## BAtch

```python
responce = models.batch(["What is the capital of France?", "What is AI?","what is quantom computing?"])

for responce in responce:
    print(responce.content)
```

```python
test = models.invoke("What is the capital of France?")
test.content.upper()
```

```python
from langchain.tools import tool
@tool
def get_weather(location:str)->str:
    """get the weather at a location"""

    return f"The weather in {location} is sunny."


model_with_tools = models.bind_tools([get_weather])
```

```python
responce = model_with_tools.invoke("what is weather in debai")

for tool_call in responce.tool_calls:
    print(f"tool:{tool_call['name']}")
    print(f"arguments:{tool_call['args']}")
```

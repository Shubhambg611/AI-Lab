```python
from fastapi import FastAPI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI
import os
from dotenv import load_dotenv
load_dotenv()
from langserve import add_routes
import openai
from langchain_groq import ChatGroq
model = ChatGroq(groq_api_key=os.getenv("groq_api_key"), model_name="groq/compound-mini")
```

## message history
Msg history” means message history: the record of messages previously sent and received in a chat or conversation. It provides context so later replies can refer to earlier ones.



```python
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

store = {}

def get_session_history(session_id: str) -> BaseChatMessageHistory:
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]


with_message_history = RunnableWithMessageHistory(model, get_session_history)
```

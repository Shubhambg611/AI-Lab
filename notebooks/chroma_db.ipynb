# ChromaDB is an open-source vector database commonly used in AI and Retrieval-Augmented Generation (RAG) applications.
It stores:
- Text documents or chunks
- Their vector embeddings
- Metadata and IDs

```python
from langchain_chroma import Chroma
from langchain_community.document_loaders import TextLoader
from langchain_community.embeddings import OpenAIEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader = TextLoader("/storage/nvme/projects/AI-Lab/downloads/datasets/speech.txt")
```

```python
data = loader.load()
```

```python
data
```

```python
text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
```

```python
splits = text_splitter.split_documents(data)
splits
```

```python
embeddings = OpenAIEmbeddings()
vectordb = Chroma.from_documents(splits, embedding=embeddings)
vectordb
```

```python
##query this db
```

```python
query = "What is the main topic of the text?"
docs = vectordb.similarity_search(query)
docs
```

```python
vectordb = Chroma.from_documents(documents=splits, embedding=embeddings,persist_directory="/storage/nvme/projects/AI-Lab/notebooks")
```

```python
docs = vectordb.similarity_search(query)
```

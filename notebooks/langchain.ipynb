```python
from langchain_community.document_loaders import WebBaseLoader
import bs4
web = WebBaseLoader(web_paths=("https://lilianweng.github.io/posts/2026-07-04-harness/",),bs_kwargs=dict(parse_only=bs4.SoupStrainer(class_=("post-title","post-content","post-hearder"))))
web.load()
```
